```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import scipy.stats as stats
```

```python
def gaussian_kl(pred_mu, pred_var):
    return 0.5 * (pred_mu.pow(2) + pred_var - 1 - torch.log(pred_var)) 

def gaussian_entropy(var):
    return 0.5*torch.log(2*torch.pi*torch.exp*var)
```

```python
p_mu = torch.zeros(1,)
p_var = torch.ones(1,)*2

target_mu = torch.zeros(1,)
target_var = torch.ones(1,)

x = np.linspace(-5,5,250)
plt.plot(x, stats.norm.pdf(x, p_mu, p_var), 'b', label='pred')
plt.plot(x, stats.norm.pdf(x, target_mu, target_var), 'k',label='target')
plt.legend()
plt.show()

print('kld N(mu, var) || N(0,1) -> ',gaussian_kl(p_mu, p_var).item())
```
