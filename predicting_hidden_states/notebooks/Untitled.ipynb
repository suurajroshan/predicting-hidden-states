```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm 
```

```python
x = np.linspace(-5,5,500)

standard_gaussian = norm.pdf(x, 0, 1)
peaky_gaussian = norm.pdf(x, 0, 0.2)

plt.plot(x, standard_gaussian)
plt.plot(x, peaky_gaussian, 'r')
plt.show()
```

```python
def kl_gaussian(mu1, sigma1, mu2, sigma2):
    """KL( N(mu1, sigma1^2) || N(mu2, sigma2^2) )"""
    return np.log(sigma2 / sigma1) + (sigma1**2 + (mu1 - mu2)**2) / (2 * sigma2**2) - 0.5
```

```python
x = np.linspace(-5,5,500)

mu = 0
std = 0.2

standard_gaussian = norm.pdf(x, 0, 1)
peaky_gaussian = norm.pdf(x, mu, std)

plt.plot(x, standard_gaussian)
plt.plot(x, peaky_gaussian, 'r')
plt.show()
kl_gaussian(mu, std, 0 ,1).item()
```
