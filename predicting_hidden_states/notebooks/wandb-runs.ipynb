```python
import wandb
import glob 
import os

api = wandb.Api()
```

```python
runs = api.runs("hidden-state-predictions/llama-3B-gumbel-layers")
```

```python
all_strings = [] 
for run in runs:
    if 'oldcluster' in run.name:
        continue
    model_path = os.path.join('/home/woody/iwi5/iwi5368h/models/llama_3B_PHi', f'*{run.id}')
    _exists = glob.glob(model_path)
    all_strings.append(f'{run.name}, {run.id} exists?: {_exists!=[]}')
sorted(all_strings)
```
